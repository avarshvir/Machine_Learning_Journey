```python
import pandas as pd
```

```python
df = pd.DataFrame([
    [25000,200,30.5],
    [18000,150,12.5],
    [9000,100,25.7],
    [40000,300,21.2]
    ],
    columns=['A','B','C']
                 )
```

```python
df
```

```python
df_copy = df.copy()
```

```python
df_copy
```

```python
for col in df_copy.columns:
    df_copy[col] = (df[col] - df[col].mean()) / df[col].std()
```

```python
df_copy
```

```python
df_copy2 = df.copy()
```

```python
df_copy2
```

```python
for col in df_copy2.columns:
    df_copy2['A'] = (df_copy2['A'] - df_copy2['A'].mean()) / df_copy2['A'].std()
```
