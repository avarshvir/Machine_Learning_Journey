```python
import pandas as pd
```

```python
data = pd.DataFrame([
    [25000,200,30.5],
    [18000,150,12.5],
    [9000,100,25.7],
    [40000,300,21.2]
    ],
    columns=['A','B','C']
                 )
```

```python
data
```

```python
data2 = data.copy()
```

```python
data2
```

```python
for col in data2.columns:
    data2[col] = (data[col] - data[col].min()) / (data[col].max() - data[col].min())
```

```python
data2
```

```python
data2.plot(kind = 'bar')
```

```python
data3 = data.copy()
```

```python
data3
```

```python
for col in data3['A']:
    data3['A'] = (data['A'] - data['A'].min()) / (data['A'].max() - data['A'].min())
```

```python
data3
```
