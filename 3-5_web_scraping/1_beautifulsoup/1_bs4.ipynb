## Instagram BS4

```python
from bs4 import BeautifulSoup
```

```python
from urllib.request import urlopen #for handling url handling, url read, and url opening
```

```python
handle = "india"
url="https://www.instagram.com/"
page=urlopen(url+handle).read()

soup = BeautifulSoup(page,"html.parser")
print(soup)
```

```python
print(soup.prettify())
```

```python
string = soup.find("meta", property="og:description")['content']
```

```python
print(string)
```
