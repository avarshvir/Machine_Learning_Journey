```python
import pandas as pd
```

```python
data = pd.read_csv('/content/iris_dataset.csv')
```

```python
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
data.dropna(inplace=True)
```

```python
data.info()
```

```python
data.drop('Species', axis=1, inplace=True)
```

```python
data.drop('Id', axis=1, inplace=True)
```

```python
data.head()
```

```python
from sklearn.cluster import KMeans
```

```python
model = KMeans(n_clusters=3)
```

```python
data.columns
```

```python
target = model.fit_predict(data.loc[:,['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']])
```

```python
target
```

```python
data.loc[:,'target']=target
```

```python
data.head()
```

```python
model.cluster_centers_
```

```python
model.labels_
```

```python
data[data.loc[:,'target']==0].SepalLengthCm
```

```python
center= model.cluster_centers_
```

```python
center
```

```python
center[:,0]
```
