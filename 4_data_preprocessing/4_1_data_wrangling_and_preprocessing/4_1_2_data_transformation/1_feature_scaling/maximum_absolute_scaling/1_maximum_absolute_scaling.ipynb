```python
import pandas as pd
```

```python
df = pd.DataFrame([
    [25000,200,30.5],
    [18000,150,12.5],
    [9000,100,25.7],
    [40000,300,21.2]
    ],
    columns=['A','B','C']
                 )
```

```python
df
```

```python
import matplotlib.pyplot as plt
```

```python
df.plot(kind='bar')
```

```python
df_copy = df
```

```python
df_copy
```

```python
for col in df_copy.columns:
    df_copy[col]=df_copy[col]/df_copy[col].abs().max()
```

```python
df2 = pd.DataFrame([
    [25000,200,30.5],
    [18000,150,12.5],
    [9000,100,25.7],
    [40000,300,21.2]
    ],
    columns=['A','B','C']
                 )
```

```python
mas = MaxAbsScaler()
```

```python
scaled_A = mas.transform(df2[['A']])
```
