## Removing Duplicate Values

```python
import pandas as pd
```

```python
data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'David'],
    'Age': [25, 30, 25, 40],
    'City': ['NY', 'LA', 'NY', 'SF']
})
```

```python
data
```

```python
data = data.drop_duplicates(subset=['Name', 'Age'], keep='first')
```
