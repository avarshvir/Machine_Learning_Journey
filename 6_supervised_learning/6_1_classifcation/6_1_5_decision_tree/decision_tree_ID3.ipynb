```python
import pandas as pd
```

```python
data = pd.read_csv('/content/social_ads.csv')
```

```python
data
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
X = data.iloc[:,:-1]
```

```python
y = data.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
X_train,X_test, y_train, y_test = tts(X,y,train_size=0.75, random_state=44)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
X_train = sc.fit_transform(X_train)
```

```python
X_test = sc.fit_transform(X_test)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtid3 = DecisionTreeClassifier(criterion='entropy',random_state = 0)
```

```python
dtid3.fit(X_train, y_train)
```

```python
y_pred = dtid3.predict(X_test)
```

```python
y_pred
```

```python
y_test.values
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, precision_score, recall_score, f1_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test, y_pred)
```

```python
f1_score(y_test, y_pred)
```
