```python
import pandas as pd
```

```python
data = pd.read_csv('/content/social_ads.csv')
```

```python
data
```

```python
X = data.iloc[:,:-1].values
```

```python
y = data.iloc[:,-1].values
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
X_train,X_test, y_train, y_test = tts(X,y, train_size=0.8, random_state=44)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dtcls = DecisionTreeClassifier(criterion='gini', splitter='random',random_state=0)
#dtcls = DecisionTreeClassifier()
```

```python
dtcls.fit(X_train, y_train)
```

```python
y_pred = dtcls.predict(X_test)
```

```python
y_test
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score, precision_score, f1_score
```

```python
accuracy_score(y_test, y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
recall_score(y_test, y_pred)
```

```python
precision_score(y_test, y_pred)
```

```python
f1_score(y_test, y_pred)
```
