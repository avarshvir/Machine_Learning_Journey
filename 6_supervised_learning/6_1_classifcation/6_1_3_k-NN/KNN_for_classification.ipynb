# KNN Algorithm for classification problem
### choosing k = underroot of n which is 23, k = 23
### choosing using GridSearchCV k = 3
- when k = 23, accuracy is ~96
- when k = 3, accuracy is ~95

```python
import pandas as pd
```

```python
data = pd.read_csv('/content/KNNAlgorithmDataset.csv')
```

```python
data.info()
```

```python
data.drop('id',axis = 1, inplace = True)
```

```python
data.drop('Unnamed: 32', axis = 1, inplace = True)
```

```python
data['diagnosis'].value_counts()
```

```python
data['diagnosis'].unique()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
data['diagnosis'] = le.fit_transform(data.iloc[:,0])
```

```python
data
```

```python
X = data.iloc[:,1:]
```

```python
X
```

```python
y = data.iloc[:,0]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
X_train, X_test, y_train, y_test = tts(X,y, train_size= 0.8, random_state= 44)
```

```python
print("X_train.shape:",X_train.shape, "X_test.shape:",X_test.shape, "y_train.shape:",y_train.shape, "y_test.shape:",y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model = KNeighborsClassifier(n_neighbors= 23)
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test.values
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, recall_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
recall_score(y_test,y_pred)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
params = {
    'n_neighbors': [i for i in range(1, 31)]
}
```

```python
model2 = KNeighborsClassifier()
```

```python
cv= GridSearchCV(model2,param_grid=params, scoring='accuracy')
```

```python
cv.fit(X_train,y_train)
```

```python
cv.best_estimator_
```

```python
model3 = KNeighborsClassifier(n_neighbors=3)
```

```python
model3.fit(X_train, y_train)
```

```python
y_pred_m3 = model3.predict(X_test)
```

```python
y_pred_m3
```

```python
y_test.values
```

```python
accuracy_score(y_test, y_pred_m3)
```

```python
confusion_matrix(y_test, y_pred_m3)
```
