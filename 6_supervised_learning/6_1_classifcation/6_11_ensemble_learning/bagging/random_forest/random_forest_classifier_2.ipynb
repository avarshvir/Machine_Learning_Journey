```python
import pandas as pd
```

```python
data = pd.read_csv('/content/iris.csv')
```

```python
data
```

```python
data['sepal_width'].fillna(data['sepal_width'].mean(), inplace=True)
```

```python
data.isnull().sum()
```

```python
data['petal_width'].fillna(0.2,inplace= True)
```

```python
data
```

```python
data['species'].fillna('setosa', inplace=True)
```

```python
data
```

```python
X = data.iloc[:,:-1]
```

```python
X
```

```python
y = data.iloc[:,-1]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
X_train, X_test, y_train, y_test = tts(X, y, train_size=0.8, random_state = 44)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators=50, criterion='gini')
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test.values
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, f1_score, recall_score, precision_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

```python
recall_score(y_test, y_pred,average='weighted')
```

```python
f1_score(y_test, y_pred, average='weighted')
```

```python
precision_score(y_test, y_pred, average = 'weighted')
```
