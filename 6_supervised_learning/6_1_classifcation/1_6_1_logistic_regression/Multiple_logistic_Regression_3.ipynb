```python
import pandas as pd
```

```python
data = pd.read_csv('/content/iris.csv')
```

```python
data.isnull().sum()
```

```python
data['sepal_width'].fillna(data['sepal_width'].mean(), inplace = True)
```

```python
data['petal_width'].fillna(data['petal_width'].mean(), inplace = True)
```

```python
data.isnull().sum()
```

```python
data['species'].fillna('setosa', inplace = True)
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
labels = le.fit_transform(data['species'])
```

```python
data['species'] = labels
```

```python
data['species'].unique()
```

```python
X = data.iloc[:,:4]
```

```python
X
```

```python
y = data.loc[:,['species']]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split as tts
```

```python
X_train,X_test, y_train, y_test = tts(X,y, test_size = 0.2, random_state = 44)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, precision_score, recall_score
```

```python
ac = accuracy_score(y_test, y_pred)
```

```python
ac
```

```python
cm = confusion_matrix(y_test, y_pred)
```

```python
cm
```

```python
fs = f1_score(y_test, y_pred, average = 'weighted')
```

```python
ps = precision_score(y_test, y_pred, average = 'weighted')
```
