```python
import numpy as np
import pandas as pd
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
X = np.array([[0, 1], [1, 1], [1, 0], [0, 0]])
y = np.array([0, 1, 2, 0])  # 3 classes: 0, 1, 2
```

```python
X
```

```python
y
```

```python
# create a NN
model = Sequential()
model.add(Dense(10, input_dim=2, activation='relu'))  # Hidden layer
model.add(Dense(3, activation='softmax'))  # Output layer
```

```python
# computing the model
model.compile(loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

```python
# train the model
model.fit(X, y, epochs=10)
```

```python
# Make predictions
predictions = model.predict(X)
print(predictions)  # Output will be probabilities for each class
```
