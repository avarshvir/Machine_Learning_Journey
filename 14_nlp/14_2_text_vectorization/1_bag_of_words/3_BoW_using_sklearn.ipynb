```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
vectorizer = CountVectorizer()
```

```python
documents = [
    "I love dogs and I love cats",
    "I hate snakes but love animals",
    "Dogs and cats are great pets"
]
```

```python
bow_matrix = vectorizer.fit_transform(documents)
```

```python
bow_matrix
```

```python
bow_matrix.toarray()
```

```python
print("Vocabulary:", vectorizer.get_feature_names_out())
```

```python
bow_matrix.shape
```

```python
vectorizer_advanced = CountVectorizer(
    lowercase=True,
    stop_words='english',  # remove common English stop words
    ngram_range=(1, 2),    # include unigrams and bigrams
    max_features=100       # limit vocabulary size
)
```

```python
X_advanced = vectorizer_advanced.fit_transform(documents)
print("\nAdvanced Scikit-learn:")
print("Vocabulary size:", len(vectorizer_advanced.get_feature_names_out()))
print("Feature names:", vectorizer_advanced.get_feature_names_out())
```
