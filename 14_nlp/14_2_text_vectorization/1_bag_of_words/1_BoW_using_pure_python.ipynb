```python
corpus = [
    "I love NLP and NLP",
    "NLP is powerful"
]
```

```python
corpus
```

```python
vocab = set()
```

```python
for sentence in corpus:
  for word in sentence.lower().split():
    vocab.add(word)
```

```python
vocab = sorted(list(vocab))
```

```python
vocab
```

```python
def bow_vector(text, vocab):
  words = text.lower().split()
  return [words.count(word) for word in vocab]
```

```python
for sentence in corpus:
    print(f"{sentence}: {bow_vector(sentence, vocab)}")
```
