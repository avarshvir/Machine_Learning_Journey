```python
from gensim.models import Word2Vec
```

```python
sentences = [
    ["the", "quick", "brown", "fox"],
    ["jumped", "over", "the", "lazy", "dog"],
    ["i", "love", "machine", "learning"],
    ["nlp", "is", "fun"]
]
```

```python
sentences
```

```python
# Train Word2Vec model (skip-gram by default)
model = Word2Vec(sentences, vector_size=50, window=2, sg=1, min_count=1)
```

```python
model
```

```python
# Get vector for a word
print(model.wv['fox'])
```

```python
# Find similar words
print(model.wv.most_similar('fox'))
```

```python
# CBOW: set sg=0
cbow_model = Word2Vec(sentences, vector_size=50, window=2, sg=0, min_count=1)
```

```python
cbow_model.wv['fox']
```

```python
# Better training data
good_sentences = [
    ["the", "quick", "brown", "fox", "jumped", "over", "the", "lazy", "dog"],
    ["foxes", "are", "clever", "animals", "that", "hunt", "at", "night"],
    ["wolves", "and", "foxes", "are", "both", "wild", "canines"],
    ["the", "red", "fox", "ran", "through", "the", "forest"],
    ["dogs", "and", "foxes", "share", "common", "ancestors"],
    ["the", "arctic", "fox", "has", "white", "fur", "in", "winter"],
    ["fox", "hunting", "is", "a", "controversial", "sport"],
    ["the", "foxy", "person", "was", "very", "clever", "and", "sly"]
]
```

```python
# Train on good data
good_model = Word2Vec(good_sentences, vector_size=50, window=2, min_count=1, epochs=100, negative=10)
```

```python
print("=== WITH GOOD TRAINING DATA ===")
print("Most similar to 'fox':")
print(good_model.wv.most_similar('fox', topn=5))
```

```python
print("\nMost similar to 'dog':")
print(good_model.wv.most_similar('dog', topn=5))
```
