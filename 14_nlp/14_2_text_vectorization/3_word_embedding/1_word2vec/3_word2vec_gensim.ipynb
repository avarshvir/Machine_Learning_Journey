```python
# Import
from gensim.models import Word2Vec, KeyedVectors
import gensim.downloader as api
```

```python
# Your data (list of tokenized sentences)
sentences = [
    ["king", "is", "a", "male", "ruler"],
    ["queen", "is", "a", "female", "ruler"],
    ["man", "is", "male", "human"],
    ["woman", "is", "female", "human"]
]

# Train model
model = Word2Vec(
    sentences=sentences,
    vector_size=100,    # Embedding dimension
    window=5,           # Context window size
    min_count=1,        # Ignore words with freq < 1
    sg=1,               # 1=Skip-gram, 0=CBOW
    workers=4,          # Parallel processing
    epochs=100          # Training iterations
)
```

```python
# Get vector for a word
vector = model.wv['king']
print(f"Vector shape: {vector.shape}")  # (100,)

# Check if word exists
if 'computer' in model.wv:
    print("Word exists in vocabulary")
```

```python
vector
```

```python
# Most similar words
similar_words = model.wv.most_similar('king', topn=3)

# Similarity between two words
similarity = model.wv.similarity('king', 'queen')
print(f"Similarity: {similarity:.3f}")
```

```python
similar_words
```

```python
# king - man + woman = queen
result = model.wv.most_similar(
    positive=['king', 'woman'],
    negative=['man'],
    topn=3
)
# Output: [('queen', 0.79), ...]
```

```python
result
```

```python
# Find odd one out
odd_one = model.wv.doesnt_match(['king', 'queen', 'apple', 'prince'])
# Output: 'apple'

# Get vocabulary
vocab = list(model.wv.key_to_index.keys())
print(f"Vocabulary size: {len(vocab)}")
```

```python
# Save complete model (can resume training)
model.save("my_word2vec.model")

# Save only word vectors (smaller, for inference)
model.wv.save("word_vectors.kv")

# Save in original Word2Vec format
model.wv.save_word2vec_format("vectors.txt", binary=False)
```
