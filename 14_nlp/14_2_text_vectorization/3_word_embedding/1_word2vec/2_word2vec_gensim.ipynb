```python
corpus = [
    ["the", "quick", "brown", "fox"],
    ["jumped", "over", "the", "lazy", "dog"],
    ["i", "love", "machine", "learning"],
    ["nlp", "is", "fun"]
]
```

```python
from gensim.models import Word2Vec
```

```python
model = Word2Vec(
    sentences=corpus,
    vector_size=50,    # Dimensionality of vectors
    window=2,          # Context window size
    sg=1,              # 1=skip-gram; 0=CBOW
    min_count=1,       # Ignores words with total frequency < min_count
    workers=2,         # For multicore machines
    seed=42,            # For reproducibility
    epochs = 10
)
```

```python
print(model.wv["fox"])  # Prints the vector for "fox" (shape: [50,])
```

```python
print(model.wv.most_similar("fox"))  # Returns words closest to "fox"
```

```python
sim = model.wv.similarity("dog", "fox")  # Cosine similarity between "dog" and "fox"
```

```python
# saving and loading model
# Save model
model.save("my_word2vec.model")

# Load model
from gensim.models import Word2Vec
model = Word2Vec.load("my_word2vec.model")
```
