```python
nlp = spacy.load('en_core_web_sm')
```

```python
text = "Elon Musk founded SpaceX in 2002 and acquired Twitter for $44 billion in 2022."
```

```python
text
```

```python
doc = nlp(text)
```

```python
doc
```

```python
for ent in doc.ents:
    print(ent.text,"->", ent.label_,":",spacy.explain(ent.label_))
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```
