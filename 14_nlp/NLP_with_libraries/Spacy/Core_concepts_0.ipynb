```python
from spacy.matcher import Matcher
```

```python
nlp = spacy.load('en_core_web_md')
```

```python
doc = nlp("Hello, world! This is spaCy.")
```

```python
for token in doc:
    print("token_text: ",token.text,"token_pos: ", token.pos_,"token_dep: ", token.dep_)
```

```python
for token in doc:
    print("token_text: ",token.text,"token_pos: ", spacy.explain(token.pos_),"token_dep: ", spacy.explain(token.dep_))
```
