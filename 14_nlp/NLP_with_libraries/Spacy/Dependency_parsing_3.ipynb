```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp("He went to play basketball.")
```

```python
doc
```

```python
for token in doc:
    print(token.text,"---->", token.dep_)
```

```python
for token in doc:
    print(token," ---> ", spacy.explain(token.dep_))
```
