```python
nlp = spacy.load('en_core_web_md')
```

```python
text = """Hey there! I'am Arshvir and I'm studying B.Tech in Computer Science and Engineering from Punjab, India. I'm always eager to learn new things and continue exploring new Technologies and different areas of Knowledge. I have a good grasp on Programming, and I'm also familiar with basic Excel and Photoshop skills.
Apart from my studies, I'm also a YouTuber with a Subscriber base of 14,000+ You can check out my channel using the link I've provided click here to visit my channel. I have also a few hobbies that I really enjoy. I love gardening, listening to music, and singing.
Additionally, I also like Electronics and learning Embedded System, I have created many projects,which you can see by scrolling through my Portfolio.
I'm a hardworlking and dedicated individual who is commited to self -improvement, and I accepted that I am an Introvert but it seems that Introvertness is my power😜"""
```

```python
text
```

```python
doc = nlp(text)
```

```python
doc
```

```python
for ent in doc.ents:
    print(ent.text,"->",ent.label_,":",spacy.explain(ent.label_))
```

```python
from spacy import displacy

displacy.render(doc, style="ent", jupyter=True)
```
