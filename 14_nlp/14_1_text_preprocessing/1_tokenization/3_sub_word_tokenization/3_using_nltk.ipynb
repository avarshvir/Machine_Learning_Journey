```python
from nltk.tokenize import RegexpTokenizer
```

```python
text = "Let's tokenize unbreakable and replaying words"
```

```python
tokenizer = RegexpTokenizer(r'\w+|un|re|ing|able')
```

```python
print(tokenizer.tokenize(text))
```
