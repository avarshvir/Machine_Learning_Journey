```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
def spacy_sentence_tokenizer(text):
  doc = nlp(text)
  return [sent.text.strip() for sent in doc.sents]
```

```python
text = "Artificial Intelligence is fascinating. Machine learning is a part of AI."
print(spacy_sentence_tokenizer(text))
```
