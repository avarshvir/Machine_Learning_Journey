# SnowballStemmer, LancasterStemmer, Lovins Stemmer

### Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
stemmer = SnowballStemmer("english")
```

```python
words = ["running", "fishing", "happily", "studies"]
```

```python
for word in words:
    print(f"Original: {word} --> Stemmed: {stemmer.stem(word)}")
```

### Lancaster Stemmer

```python
from nltk.stem import LancasterStemmer
```

```python
stemmer = LancasterStemmer()
```

```python
words = ["maximum", "presumably", "running", "happiness"]
```

```python
for word in words:
    print(f"Original: {word} --> Stemmed: {stemmer.stem(word)}")
```
