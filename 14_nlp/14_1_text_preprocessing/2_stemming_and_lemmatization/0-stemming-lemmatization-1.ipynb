# Stemming and Lemmatization

### Stemming
- **from nltk.stem import PorterStemer**

```python
import nltk
```

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
```

```python
text = "NLTK includes several stemming algorithms running, runner, runs, easily, fairer."
```

```python
text
```

```python
words = word_tokenize(text)
```

```python
ps = PorterStemmer()
```

```python
stemmed = [ps.stem(word) for word in words]
```

```python
stemmed
```

```python
stemmed_words = []
```

```python
for word in words:
    stemmed_words.append(ps.stem(word))
```

```python
stemmed_words
```

### Lemmatization
- **from nltk.stem import WordNetLemmatizer**
- **from nltk.corpus import wordnet**

- *WordNet is like a massive English dictionary + thesaurus created by Princeton University. Think of it as:
A giant word database containing ~200,000 words
Words are grouped by meaning (synonym sets called "synsets")
Shows relationships between words (hypernyms, hyponyms, etc.)
Contains: nouns, verbs, adjectives, adverbs*

- *WordNetLemmatizer is a tool that uses WordNet to find the base/dictionary form (lemma) of words.*

How wordnetlemmatizer works:
- - Takes a word like "running"
  - Looks it up in WordNet database
  - Returns the base form: "run"*

```python
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
text = "The children are running quickly and the mice were eating cheese."
```

```python
words = word_tokenize(text)
```

```python
words
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatized = [lemmatizer.lemmatize(word) for word in words]
```

```python
lemmatized
```
