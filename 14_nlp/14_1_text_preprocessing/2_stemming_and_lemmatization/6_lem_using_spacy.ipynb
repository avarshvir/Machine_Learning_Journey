spacy is useful and easier to use when implementing Lemmatization.

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
nlp
```

```python
text = "The better students were studying and waited patiently."
```

```python
doc = nlp(text)
```

```python
doc
```

```python
for token in doc:
  print(f"{token.text:<10} -> {token.lemma_}")
```

```python
for token in doc:
    print(f"'{token.text}' → '{token.lemma_}' (POS: {token.pos_})")
```
