WordNetLemmitizer is actual lemmatizer tool that use wordnet and reduce the word into it base form.

```python
from nltk.stem import WordNetLemmatizer
```

wordnet is like a massive library which have base words and mapping

```python
from nltk.corpus import wordnet
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize('better', pos=wordnet.ADJ)
```

```python
lemmatizer.lemmatize('running', pos=wordnet.VERB)
```

```python
lemmatizer.lemmatize('cars',pos='n')
```

```python
from nltk.tokenize import word_tokenize
```

```python
text = "I am learning NLP. I am learning how to tokenize words and lemmatize. I am better and going to school. I have multiple cars."
```

```python
words = word_tokenize(text)
```

```python
lemmatized = [lemmatizer.lemmatize(word) for word in words]
```

```python
lemmatized
```

```python
test_words = ['learning', 'better', 'going', 'cars', 'am']
```

```python
print("WORD\t\tAS NOUN\t\tAS VERB")
print("-" * 40)
for word in test_words:
    as_noun = lemmatizer.lemmatize(word)  # Default: pos='n'
    as_verb = lemmatizer.lemmatize(word, pos='v')
    print(f"'{word}'\t\t'{as_noun}'\t\t'{as_verb}'")
```
