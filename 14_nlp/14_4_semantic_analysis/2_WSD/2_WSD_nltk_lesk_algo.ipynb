```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize
import nltk
```

```python
nltk.download('wordnet')
nltk.download('punkt_tab')
nltk.download('punkt')
```

```python
sentence1 = "I deposited money in the bank."
sentence2 = "He sat on the river bank and watched the sunset."
```

```python
sense1 = lesk(word_tokenize(sentence1), 'bank')
sense2 = lesk(word_tokenize(sentence2), 'bank')
```

```python
print("Sense in sentence 1:", sense1)
print("Definition:", sense1.definition())
```

```python
print("Sense in sentence 2:", sense2)
print("Definition:", sense2.definition())
```
