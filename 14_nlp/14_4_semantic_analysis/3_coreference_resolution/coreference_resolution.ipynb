```python
import spacy
nlp = spacy.load("en_coreference_web_trf")
doc = nlp("Arshvir built a chatbot and deployed it. He was excited about its performance.")
for cluster in doc.spans.get("coref_clusters", []):
    print([span.text for span in cluster])
```

```python
from transformers import pipeline
coref_pipeline = pipeline("coreference-resolution", model="rcdwl/coref-spanbert-large")
text = "Arshvir built a chatbot and deployed it. He was excited about its performance."
print(coref_pipeline(text))
```
